# Water Quality Data Preprocessing

This notebook focuses on preparing the Kerala water quality dataset
for further analysis and pollution-risk prediction.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/kerala_water_quality.csv")

df.head()
```

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
df_clean["Data Acquisition Time"] = pd.to_datetime(
    df_clean["Data Acquisition Time"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace(".", "_", regex=False)
)

df_clean.columns
```

```python
missing_values = df_clean.isnull().sum()

missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean) * 100
).round(2)

missing_percentage[missing_percentage > 0].sort_values(
    ascending=False
)
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
df_clean = df_clean.drop(columns=["Chemical_Oxygen_Demand_(mg/L)"])
```

```python
measurement_columns = [
    "Biochemical_Oxygen_Demand_(mg/L)",
    "Fecal_Coliform_(MPN/100mL)",
    "Total_Coliform_(MPN/100mL)"
]

for column in measurement_columns:
    df_clean.loc[df_clean[column] < 0, column] = np.nan
```
